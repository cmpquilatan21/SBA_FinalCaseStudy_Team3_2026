```python
import pandas as pd
import numpy as np
df = pd.read_csv("/content/SBAnational.csv")
```

```python
print("Shape (rows, cols):", df.shape)
print("\nColumns:", df.columns)
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna(subset=["MIS_Status"])
df['MIS_Status'].value_counts(dropna=False)
```

```python
# Create Franchise before dropping FranchiseCode
df["Franchise"] = (
    pd.to_numeric(df["FranchiseCode"], errors="coerce")
    .fillna(0)
    .ne(0)
    .astype(int)
)
```

```python
# Remove unneeded rows
cols_drop = [
    "Name",
    "City",
    "Zip",
    "Bank",
    "ChgOffDate",
    "ChgOffPrinGr",
    "BalanceGross",
    "CreateJob",
    "RetainedJob"
]

df = df.drop(columns=cols_drop)
```

```python
# Money text ("$60,000.00 ") -> numbers
for c in ['GrAppv', 'SBA_Appv']:
    df[c] = pd.to_numeric(df[c].astype(str).str.replace(r'[$,\s]', '', regex=True))

# Outcome: 1 = charged off, 0 = paid in full (drop only the loans with no outcome)
df = df[df['MIS_Status'].isin(['P I F', 'CHGOFF'])].copy()
df['Default'] = (df['MIS_Status'] == 'CHGOFF').astype(int)
```

```python
# LowDoc shows whether the loan used limited documentation
df["LowDoc"] = df["LowDoc"].where(
    df["LowDoc"].isin(["Y", "N"]),
    "Unknown"
)

# State is the borrower's business state
# BankState is the bank's state
# StateMatch equals 1 when both are in the same state
state_known = df["State"].notna() & df["BankState"].notna()

df["StateMatch"] = (
    (df["State"] == df["BankState"]) & state_known
).astype(int)

# Replace missing state values with Unknown
df["State"] = df["State"].fillna("Unknown")
df["BankState"] = df["BankState"].fillna("Unknown")

# NewExist identifies whether the business is new or already existing
df["NewExist"] = (
    df["NewExist"]
    .map({1.0: "Existing", 2.0: "New"})
    .fillna("Unknown")
)

# UrbanRural identifies whether the business is urban or rural
df["UrbanRural"] = (
    df["UrbanRural"]
    .map({1: "Urban", 2: "Rural"})
    .fillna("Undefined")
)

# NAICS identifies the business industry
# Extract the first two digits to create the industry sector
naics_number = pd.to_numeric(df["NAICS"], errors="coerce")

df["NAICS_Sector"] = (
    naics_number.fillna(0)
    .astype(int)
    .astype(str)
    .str.zfill(6)
    .str[:2]
)

df["NAICS_Sector"] = df["NAICS_Sector"].replace(
    "00", "Unknown"
)

# Transform the approved loan amount to reduce extreme values
df["log_GrAppv"] = np.log1p(df["GrAppv"])

# Calculate the portion of the loan guaranteed by the SBA
df["SBA_Portion"] = df["SBA_Appv"] / df["GrAppv"]
df["SBA_Portion"] = (
    df["SBA_Portion"]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)
# Clean RevLineCr
df["RevLineCr"] = df["RevLineCr"].where(
    df["RevLineCr"].isin(["Y", "N"]),
    "Unknown"
)

# Create franchise indicator before dropping FranchiseCode
df["Franchise"] = (
    pd.to_numeric(df["FranchiseCode"], errors="coerce")
    .fillna(0)
    .ne(0)
    .astype(int)
)
```

```python
# Convert all required money fields
for column in ["DisbursementGross", "GrAppv", "SBA_Appv"]:
    df[column] = pd.to_numeric(
        df[column].astype(str).str.replace(r"[$,\s]", "", regex=True),
        errors="coerce"
    )

# Parse dates before dropping them
df["ApprovalDate"] = pd.to_datetime(
    df["ApprovalDate"], errors="coerce"
)

df["DisbursementDate"] = pd.to_datetime(
    df["DisbursementDate"], errors="coerce"
)
```

```python
print("Shape:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
print(df.isnull().sum().sort_values(ascending=False).head())
```
